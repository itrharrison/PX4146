## Question

You want to send a document to your bank and it is very important it gets there the next day. You therefore send two copies of the document using Hermes (delivery service $A$) and the Royal Mail (delivery service $B$).  The probabilities of those companies delivering on time are:

- 0.90 for service A
- 0.80 for service B 
- 0.75 for both services being on time.

Does your strategy improve the chances the document will be delivered on time? 

### Answer

Let $\mathrm{A}$ be the event that service A arrives on time.\
Let $\mathrm{B}$ be the event that service B arrives on time.

We know that\
$P(\text{A}) = 0.9$\
$P(\text{B}) = 0.8$, and\
$P(\text{A}\cap\text{B}) = 0.75$


We are interested in the probability that either event $\mathrm{A}$ *or* $\mathrm{B}$ occurs.

From the lecture slides, we know how to calculate this:

$P(\text{A}\cup\text{B}) = P(\text{A}) + P(\text{B}) - P(\text{A}\cap\text{B})$

and the relevant probabilities are given.

So the required result of either package arriving on time is the simple calculation:

```python
p_A = 0.9
p_B = 0.8
p_both = 0.75

p_either = p_A + p_B - p_both
```

```python
print(f'Probability that either arrives on time = {p_either:.2f}')
```

This is greater than both $P(\text{A})$ and $P(\text{B})$, so this strategy does improve the probability of them receiving the document on time.

## <font color=#30a64d>Question</font>

You flip a coin 5 times. What is the probability that heads will not occur 3 times?    

### Answer

We assume that this means "exactly 3 times"

So we are interested in\
$P(\text{Heads does not occur exactly 3 times}) = 1 - P(\text{Heads occurs exactly 3 times})$

We know we have 5 flips and we want 3 to have the $\text{Heads}$ outcome, which has a $P(H)=0.5$.

Discussed explicitly in Chapter 3, but already known to us from previous stats courses(!) is the Binomial Distribution for getting $k$ successes from $n$ trials, where each success has probability $p$:\
$P(X = k) = \left(\begin{array}{c} n \\ k \end{array}\right)p^k (1 - p)^{n-k}$

For this case:
$P(3\text{Heads}) = \left(\begin{array}{c} 5 \\ 3 \end{array}\right){0.5}^{5}$

which we can calculate simply:

```python
# here I import the relevant python scipy function for the binomial coefficient (n,k)
# which is overkill for this case, but just so you know it exists!
from scipy.special import binom

p_3heads = binom(5,3) * 0.5**5
p_not3heads = 1 - p_3heads
print(p_not3heads)
```

## <font color=#30a64d>Question</font>

Suppose you think you have a rare disease where 1 in 10,000 people are affected. You go to your doctor, and she performs a test. The test is of a high quality, and is correct 99% of the time.

Unfortunately your results come back positive for the disease. What is your chance of having the disease? Is it (roughly): <br><br>

a) 99% <br>
b) 90% <br>
c) 10% or <br>
d) 1%? <br> <br>

Repeat the same as above, but change 1 in 10,000 to 1 in 100. What would the chances of having the disease be now?

### Answer

Assuming same True Positive Result (TPR) as True Negative Result (TNR) probabilities (i.e, the given $P(\text{Correct Result}) = 0.99$)
\
\
$$
\begin{align}
P(\text{Has disease}|\text{Test positive}) &= \frac{P(\text{Test positive}|\text{Has disease})*P(\text{Has disease})}{P(\text{Test positive})}\\
&= \frac{P(\text{Test positive}|\text{Has disease})*P(\text{Has disease})}{P(\text{Test positive}|\text{Has disease})*P(\text{Has disease})+P(\text{Test positive}|\text{No disease})*P(\text{No diseas e})}\\
\end{align}
$$
\
again, the python to calculate this is simple:


```python
p_disease = 1. / 10000
p_nodisease = 1 - p_disease
p_truepos = 0.99
p_falsepos = 1 - p_truepos

p_pos = p_truepos * p_disease + p_falsepos * p_nodisease

p_d_pos = p_truepos * p_disease / p_pos

print(f'{100*p_d_pos:.1f}%')
```

So even though the test is superficially very good (99% accurate!) then because of the relatively low prevelance of the disease, we still create a lot of false positives by applying the test to a random selection of people. This is why giving lots of people medical tests when they do not have other symptons (which would change the prior probability of $\text{P(Has diseases)}$) is sometimes not a useful thing to do.


Imagine we gave everyone in the UK a 99% accurate test for a rare disease. The number of false positives would be very large!

For the other case where the disease is much more prevalent we see the test becomes (a bit) more useful:

```python
p_disease = 1. / 100
p_nodisease = 1 - p_disease
p_truepos = 0.99
p_falsepos = 1 - p_truepos

p_pos = p_truepos * p_disease + p_falsepos * p_nodisease

p_d_pos = p_truepos * p_disease / p_pos

print(f'{100*p_d_pos:.1f}%')
```

For those of you thinking about coding better, this could be more nicely written in the form of a function:

```python
def p_hasdisease(p_disease, p_truepos, percent=True):
    p_nodisease = 1 - p_disease
    p_falsepos = 1 - p_truepos

    p_pos = p_truepos * p_disease + p_falsepos * p_nodisease

    p_d_pos = p_truepos * p_disease / p_pos

    if percent:
        return 100*p_d_pos
    else:
        return p_d_pos
```

```python
print(f'Case 1 (rare disease): {p_hasdisease(1. / 10000, 0.99):.1f}%')
print(f'Case 2 (common disease): {p_hasdisease(1. / 100, 0.99):.1f}%')
```

## Question

We saw in [Chapter 2 Notebook](https://github.com/itrharrison/PX4146/blob/main/Chapter2/Chapter2.ipynb) that the probability that a couple has 2 children that are boys when the older child is a boy is 50%. 

What about the following scenario: a couple has two children, of which at least one is a boy. If the probabilities of having a boy or a girl are both 50%, what is the probability that the couple has two boys?

At first glance, this appears to be asking the same question as in our previous notebook. We might reason as follows: “We know that one is a boy, so the only question is whether the other one is a boy, and the chances of that being the case are 50%. So again, the answer is 50%.”

Although this sounds common sense, it is incorrect.  Show why.

### Answer

$$
\begin{align}
P(\text{Both children are boys}|\text{At least one is a boy}) &= \frac{P(\text{At least one is a boy}|\text{Both children are boys})*P(\text{Both children are boys})}{P(\text{At least one is a boy})}\\
&= \frac{P(\text{Both children are boys})}{1-P(\text{None are boys})}
\end{align}
$$
\
This follows because $P(\text{At least one is a boy}|\text{Both children are boys})=1$\
(if both are boys, then at least one will be a boy)

```python
p_boy = 0.5
p_notboy = 1 - p_boy

p_bb_given_b = (1 * p_boy**2) / (1 - p_notboy**2)

print(f'Probability of both children being boys, given at least one is a boy = {p_bb_given_b:.2f}')
```
